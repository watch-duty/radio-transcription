# Multi-Model STT Evaluation Study

This notebook implements an orchestration pipeline for reading speech-to-text manifest files from Google Cloud Storage (GCS), downloading audio files locally, and sending them to an arbitrary set of FastAPI STT model endpoints. 

Transcriptions for all defined models are dynamically categorized and uploaded back to GCS under categorized model paths.

```python
# @title Model Evaluation Driver
# @markdown Select the model and configure the parameters.
# @markdown Note: Ensure `run_model_evaluation_study.py` is accessible.

from run_model_evaluation_study import run_pipeline

# @markdown ### GCP Configuration
GCP_PROJECT_ID = "your-project-id" #@param {type:"string"}
GCS_MANIFEST_URI = "gs://your-bucket/manifest.jsonl" #@param {type:"string"}
GCS_BUCKET = "your-bucket" #@param {type:"string"}
PROJECT_NAME = "evaluation_project" #@param {type:"string"}
EXPERIMENT_NAME = "exp_v1" #@param {type:"string"}

# @markdown ### Model Selection
SELECTED_MODEL = "Canary-Qwen-2.5B" #@param ["Qwen3-ASR", "Whisper", "Gemma3N", "Canary-Qwen-2.5B"]

# Define your endpoints here
MODEL_ENDPOINTS = {
    "Qwen3-ASR": "http://qwen-asr-cpu:8000/transcribe",
    "Canary-Qwen-2.5B": "http://canary-qwen:8000/transcribe",
    "Whisper": "http://your-whisper-endpoint:8000/transcribe",
    "Gemma3N": "http://your-gemma-endpoint:8000/transcribe"
}

PROMPT = "" #@param {type:"string"}

# @markdown ### Logging & Uploads
VERBOSE = False #@param {type:"boolean"}
NO_UPLOAD = False #@param {type:"boolean"}

# @markdown ### Testing
LIMIT = 5 #@param {type:"integer"}
# @markdown *Set to 0 to process the full manifest. Mostly used for testing.*

# Run the pipeline
run_pipeline(
    gcp_project_id=GCP_PROJECT_ID,
    gcs_manifest_uri=GCS_MANIFEST_URI,
    gcs_bucket=GCS_BUCKET,
    project_name=PROJECT_NAME,
    experiment_name=EXPERIMENT_NAME,
    model_endpoints=MODEL_ENDPOINTS,
    selected_model=SELECTED_MODEL,
    prompt=PROMPT if PROMPT else None,
    verbose=VERBOSE,
    no_upload=NO_UPLOAD,
    limit=LIMIT if LIMIT > 0 else None
)
```
